## 1. Create strings using single, double, and triple quotes.

```python
single_quoted = 'This string uses single quotes.'
double_quoted = "This string uses double quotes."
triple_quoted = """This is a triple-quoted string.
It can span multiple lines."""

print(single_quoted)
print(double_quoted)
print(triple_quoted)
```

## 2. Demonstrate indexing and negative indexing.

```python
text = "Python Programming"

print(f"Text: {text}")
print(f"text[0] = {text[0]}")
print(f"text[7] = {text[7]}")
print(f"text[-1] = {text[-1]}")
print(f"text[-5] = {text[-5]}")
```

## 3. Reverse a string using slicing.

```python
word = "Python"
reversed_word = word[::-1]

print(f"Original: {word}")
print(f"Reversed: {reversed_word}")
```

## 4. Count vowels, consonants, words, and characters.

```python
sample_text = "Hello, Python World!"

vowel_count = sum(character.lower() in "aeiou" for character in sample_text)
consonant_count = sum(
    character.isalpha() and character.lower() not in "aeiou"
    for character in sample_text
)
word_count = len(sample_text.split())
character_count = len(sample_text)  # Includes spaces and punctuation.

print(f"Text: {sample_text}")
print(f"Vowels: {vowel_count}")
print(f"Consonants: {consonant_count}")
print(f"Words: {word_count}")
print(f"Characters: {character_count}")
```

## 5. Check whether a string is a palindrome.

```python
palindrome_text = "Never odd or even"
cleaned_text = "".join(
    character.lower() for character in palindrome_text if character.isalnum()
)
is_palindrome = cleaned_text == cleaned_text[::-1]

print(f"Original text: {palindrome_text}")
print(f"Is palindrome: {is_palindrome}")
```

## 6. Compare two strings lexicographically.

```python
first_string = "apple"
second_string = "banana"

if first_string == second_string:
    print(f'"{first_string}" is equal to "{second_string}".')
elif first_string < second_string:
    print(f'"{first_string}" comes before "{second_string}".')
else:
    print(f'"{first_string}" comes after "{second_string}".')
```

## 7. Use replace(), split(), join(), and strip().

```python
sentence = "Python is easy to learn"
colors = "red,green,blue"
padded_text = "   remove extra spaces   "

color_list = colors.split(",")

print(f"replace(): {sentence.replace('easy', 'powerful')}")
print(f"split(): {color_list}")
print(f"join(): {' | '.join(color_list)}")
print(f"strip(): '{padded_text.strip()}'")
```

## 8. Format output using f-strings.

```python
name = "Aarav"
score = 92.5

print(f"{name} scored {score:.1f}% in Python.")
```

## 9. Evaluate different Boolean expressions.

```python
age = 20
has_id = True
is_raining = False
language = "Python Programming"

print(f"Age is at least 18 and has ID: {age >= 18 and has_id}")
print(f"Age is under 18 or it is raining: {age < 18 or is_raining}")
print(f"It is not raining: {not is_raining}")
print(f"'Python' is in language: {'Python' in language}")
```

## 10. Demonstrate truth-value testing with numbers, strings, lists, and dictionaries.

```python
values = {
    "number 0": 0,
    "number 42": 42,
    "empty string": "",
    "non-empty string": "Python",
    "empty list": [],
    "non-empty list": [1, 2],
    "empty dictionary": {},
    "non-empty dictionary": {"language": "Python"},
}

for description, value in values.items():
    print(f"{description}: bool({value!r}) = {bool(value)}")
```

## 11. Write a login program using Boolean logic.

```python
valid_username = "admin"
valid_password = "Python@123"

# Change these values to test another login attempt.
entered_username = "admin"
entered_password = "Python@123"

login_successful = (
    entered_username == valid_username and entered_password == valid_password
)

if login_successful:
    print("Login successful.")
else:
    print("Invalid username or password.")
```

## 12. Show short-circuit evaluation using and and or.

```python
def show_evaluation(name, value):
    print(f"Evaluated: {name}")
    return value

print("Using and:")
and_result = show_evaluation("first False value", False) and show_evaluation(
    "second True value", True
)
print(f"Result: {and_result}")

print("\nUsing or:")
or_result = show_evaluation("first True value", True) or show_evaluation(
    "second False value", False
)
print(f"Result: {or_result}")
```
