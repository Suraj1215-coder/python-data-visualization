# Assignment 4 - Operators and Conditions

The following programs use arithmetic, comparison, logical and bitwise operators.

## 1. Arithmetic operations on two numbers

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Power =", a ** b)

if b != 0:
    print("Division =", a / b)
    print("Floor division =", a // b)
    print("Remainder =", a % b)
else:
    print("Division, floor division and remainder are not possible with 0.")
```

## 2. Check whether a number is even or odd using %

```python
num = int(input("Enter an integer: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

## 3. Convert seconds into hours, minutes and seconds

```python
total_seconds = int(input("Enter total seconds: "))

hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

print(hours, "hour(s)", minutes, "minute(s)", seconds, "second(s)")
```

## 4. Calculate simple interest

```python
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time in years: "))

simple_interest = (principal * rate * time) / 100

print("Simple Interest =", simple_interest)
```

## 5. Check placement eligibility

Here, the minimum CGPA is taken as 6.5 and there should be no active backlog.

```python
cgpa = float(input("Enter CGPA: "))
backlogs = int(input("Enter number of active backlogs: "))

if cgpa >= 6.5 and backlogs == 0:
    print("Eligible for placement")
else:
    print("Not eligible for placement")
```

## 6. Check whether a year is a leap year

```python
year = int(input("Enter a year: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

## 7. Check whether a character is present in a string

```python
text = input("Enter a string: ")
character = input("Enter a character to search: ")

if character in text:
    print("Character is present in the string")
else:
    print("Character is not present in the string")
```

## 8. Difference between == and is

`==` compares values, while `is` checks whether both variables refer to the same object.

```python
list1 = [10, 20, 30]
list2 = [10, 20, 30]
list3 = list1

print("list1 == list2 :", list1 == list2)
print("list1 is list2 :", list1 is list2)
print("list1 == list3 :", list1 == list3)
print("list1 is list3 :", list1 is list3)
```

## 9. Exchange two Boolean values using XOR

```python
first = True
second = False

first = first ^ second
second = first ^ second
first = first ^ second

print("First =", first)
print("Second =", second)
```

## 10. Multiply a number by 2 using left shift

```python
num = int(input("Enter an integer: "))
result = num << 1

print("After multiplying by 2 =", result)
```

## 11. Check whether a number lies between 10 and 100

```python
num = int(input("Enter a number: "))

if 10 <= num <= 100:
    print("Number lies between 10 and 100")
else:
    print("Number does not lie between 10 and 100")
```

## 12. Calculate the final bill after discount and GST

```python
bill = float(input("Enter original bill amount: "))
discount_percent = float(input("Enter discount percentage: "))
gst_percent = float(input("Enter GST percentage: "))

discount_amount = bill * discount_percent / 100
amount_after_discount = bill - discount_amount
gst_amount = amount_after_discount * gst_percent / 100
final_bill = amount_after_discount + gst_amount

print("Discount amount =", discount_amount)
print("GST amount =", gst_amount)
print("Final bill =", final_bill)
```

## 13. Find the largest among three numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest number =", a)
elif b >= a and b >= c:
    print("Largest number =", b)
else:
    print("Largest number =", c)
```
