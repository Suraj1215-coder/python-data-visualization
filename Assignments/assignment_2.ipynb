2. Perform all arithmetic operations on two integers

```python
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

3. Convert a decimal number to binary, octal, and hexadecimal

```python
num = 20

print("Decimal:", num)
print("Binary:" , bin(num))
print("Octal:", oct(num))
print("Hexadecimal:", hex(num))
```

4. Check whether a number is even or odd

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

5. Demonstrate floor division with positive and negative integers

```python
a = 10
b = 3

print("Positive floor division:", a // b)

print("Negative floor division:", -a // b)
```

6. Convert user input from string to integer and perform arithmetic

```python
a = input("Enter first number: ")
b = input("Enter second number: ")

a = int(a)
b = int(b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

7. Implement bitwise AND, OR, XOR, left shift, and right shift

```python
a = 10
b = 5

print("AND", a&b)

print("OR", a | b)
print("XOR", a ^ b)
print("Left Shift", a << 1)
print("Right Shift", a >> 1)
```

8. Compare integer values using relational operators

```python
a = 10
b = 20

print( a == b)
print( a != b)
print(a > b)
print(a < b)
print( a >= b)
print( a <= b)
```
