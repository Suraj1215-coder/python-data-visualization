Write a program to show mutable vs immutable behavior.

```python


# Immutable
a = 10
print("Before:", a)

a = a + 5
print("After:", a)


```

```python
# Mutable
num = [30, 25, 60]
print("Before:", num)

num[1] = 100
print("After:", num)
```

Create a global variable and modify it inside a function.

```python

x = 10

def change_value():
    global x
    x = 20
    print("Inside function:", x)

print("Before function:", x)

change_value()

print("After function:", x)
```
